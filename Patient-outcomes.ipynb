# Project : Healthcare Patient Outcomes

Your hospital leadership team comes to you asking why patients keep coming back. Readmission rates across all 30 Hyderabad facilities are consistently above the 20% industry benchmark and no one can explain why. Every readmission is a signal that something in the care journey broke down, and a direct cost to the hospital.

You meet with the clinical operations lead who walks you through the full patient journey - admission, diagnosis, treatment, medication, discharge. You immediately spot the friction points: Was severity classified correctly? Was the length of stay long enough? Was the right treatment type assigned? Any one of these could be driving patients back through the door.

You then discover that patient activity lives across 8 completely separate tables that have never been joined or cleaned - orphaned records, duplicate admissions, bad discharge dates, and nulls throughout. You build 8 SQL cleaning views in HEX, model everything into a star schema with admissions at the center, and now you can finally ask the questions that matter.

What you find is more systemic than anyone expected. Every top 10 disease exceeds the readmission benchmark. Severe patients are discharged after just 4.4 days - barely 0.3 days longer than Mild patients. Diagnostic treatments carry a 21% mortality rate. And L V Prasad Eye Institute has 41% of its revenue sitting uncollected. You bring it all back to leadership through three Tableau dashboards covering patient demographics, clinical outcomes, and hospital operations - giving administrators a unified view across all 30 facilities for the first time.

The data tells a clear story: this isn't a disease problem. It's a discharge and post-care problem - and it's costing the hospitals both patients and revenue.
